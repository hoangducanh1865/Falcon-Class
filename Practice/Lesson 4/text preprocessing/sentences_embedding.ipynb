```python
from sentence_transformers import SentenceTransformer

embedder = SentenceTransformer('all-MiniLM-L6-v2')
```

```python
sentences = [
    "this is a wonderful price for the amount you get",
    "great prodect big amount",
    "i bought this for my son as his hair is thinning i do not know yet how well this is helping him"
]

embeddings = embedder.encode(sentences)
```

```python
embeddings.shape
```

```python
from sklearn.metrics.pairwise import cosine_similarity
cosine_similarity(embeddings, embeddings)[0, 1:]
```
